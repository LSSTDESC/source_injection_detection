```python
import requests
import json

```

```python
base_url = "https://api.antares.noirlab.edu/v2/"

query = {
    "query": {
        "bool": {
            "filter": [
                {
                    "range": {
                        "properties.lantern_xgboost_t2.0.7_c0.95_num_tagged_alerts": {
                            "gte": 3,
                            #"lte": 100,
                        }
                    }
                },
                {
                    "range": {
                        "properties.lantern_xgboost_t2.0.7_c0.95_max_score": {
                            "gte": 0.997,
                        }
                    }
                },
                {
                     "terms": {
                         "tags": ["lantern_xgboost_t2.0.7_c0.95"]
                     }
                }
             ]
        }
    }
}

```

```python
def query_loci(limit=100):

    endpoint = "loci"
    response = requests.get(base_url+endpoint,
                            params={"limit": limit,
                                    "elasticsearch_query": json.dumps(query)
                                   }
                           )

    response = response.json()

    return response


def query_alerts(locus_id, limit=100):

    endpoint = f"loci/{locus_id}/alerts"
    response = requests.get(base_url+endpoint,
                            params={"limit": limit,
                                    "elasticsearch_query": json.dumps({})}
                           )

    response = response.json()

    return response


def query_catalog_matches(locus_id, limit=100):

    endpoint = f"loci/{locus_id}/catalog-matches"
    response = requests.get(base_url+endpoint,
                            params={"limit": limit,
                                    "elasticsearch_query": json.dumps({})}
                           )

    response = response.json()

    return response


def get_next_response(response):
    # Link of the next page of locus ids, you can call requests.get(response["links"]["next"]) repeatedly to fetch data until next becomes None
    #print(response["links"]["next"])

    response = requests.get(response["links"]["next"])
    response = response.json()
    #print(response["data"])

    return response

```
